```python
import pandas as pd
```

```python
df_envios = pd.read_csv('envios_transporte.csv')
df_envios
```

```python
#Proyeccion y separacion de series 
columna_tipo_vehiculo = pd.Series(df_envios['Tipo_Vehiculo'])
columna_tipo_vehiculo
```

```python
#Proyeccion y separacion de series 
data = df_envios[["Ruta_Origen","Tipo_Carga","Distancia_KM"]]
df = pd.DataFrame(data)
df
```

```python
#Acceso por etiqueta
fila_consultada = df_envios.loc[99]
fila_consultada
```

```python
#Limpieza de registros duplicados
df_limpio = df_envios.drop_duplicates()
df_limpio
```

```python
#Filtrado por condicion simple 
Carga_pesada = df_limpio[df_limpio['Peso_Toneladas'] > 15.0]
Carga_pesada
```

```python
#Filtrado Categorico con .isin()
df_limpio[df_limpio['Tipo_Vehiculo'].isin(['Tractomula','Camión Mediano'])]
```

```python
#Filtrdo de Rangos numericos con .between()
df_limpio[df_limpio['Distancia_KM'].between(100.0, 250.0)]
```

```python
#Consultas con el metodo query()
df_limpio.query("Ruta_Origen == 'Pereira' & Estado_Entrega == 'Retrasado'")
```

```python
#Filtrado Logico Compuesto y Cadenas
filtro = (df_limpio["Estado_Entrega"] == "A Tiempo") & (df_limpio["Tipo_Carga"].str.startswith("E"))
df_limpio[filtro]
```

```python
#Tratamiento e imputacion de Nulos
df_llenar = df_limpio.fillna({
    "Distancia_KM" : df_limpio["Distancia_KM"].mean(),
    "Combustible_Galones" : "0.0"
})

df_llenar 
```

```python
#Columnas Calculadas con .apply()
def medicion_de_peso(peso):
    if peso < 5.0:
        return "Ligero"
    elif peso < 15.0:
        return "Intermedio"
    else:
        return "Pesado"

df_limpio["Categoria_Carga"] = df_limpio["Peso_Toneladas"].apply(medicion_de_peso) 

```
